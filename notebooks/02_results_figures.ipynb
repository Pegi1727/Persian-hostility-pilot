# Chapter 4 results figures

Regenerate the five pilot figures using the revised `scripts/make_figures.py`. This notebook does not generate the graphical abstract.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
from IPython.display import display

cwd = Path.cwd().resolve()
ROOT = next((p for p in (cwd, *cwd.parents) if (p / 'scripts/analyze.py').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Open this notebook from the repository or its notebooks/ folder.')
print(ROOT.name)
```

```python
from IPython.display import Image, Markdown
script = ROOT / 'scripts/make_figures.py'
if not script.is_file() or "OUT = ROOT / 'figures' / 'generated'" not in script.read_text(encoding='utf-8'):
    raise RuntimeError('Install the revised scripts/make_figures.py included in this package.')
run = subprocess.run([sys.executable, str(script)], cwd=ROOT,
                     check=True, capture_output=True, text=True)
print(run.stdout)
output = ROOT / 'figures/generated' 
```

```python
titles = [
    'Inter-annotator agreement for perceived hostility',
    'Perceived hostility across intended hostility conditions',
    'Perceived arousal across intended arousal conditions',
    'Perceived hostility across arousal conditions',
    'Correspondence between intended and perceived policy stance',
]
for i, title in enumerate(titles, 1):
    display(Markdown(f'### Figure 4.{i}. {title}'))
    display(Image(filename=str(output / f'Figure_4_{i}.png'), width=850))
```
